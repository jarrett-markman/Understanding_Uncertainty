### 6. Some Common Distributions

For the following distributions:

You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.


#### Support for Logistic, Exponential, Negative Binomial, and Categorical Distributions. 

The support for a distribution represents the sets of all the possible values that a random variable can take. 

- The support for the logistic distribution consists of all real numbers, which can be written as: $(-\infty, \infty)$
- The support for the exponential distribution consists of all real non-negative numbers, which can be written as: $[0, \infty)$
- The support for the negative binomial distribution consists of all read non-negative *integers*, which can be written as: $\{0, 1, 2, 3, ...\}$ for $k=0, ...n$
- The support for the categorical distribution represents the set of all possible outcomes, which can be written as: $\{1, 2, ..., k\}$ for $K$ categories. For example, when rolling a die in a categorical distribution, the support is: $\{1, 2, 3, 4, 5, 6\}$

#### Compute the PDF for the Logistic and Exponential Distributions

We are given the following CDF's for the Logistic and Exponential Distributions:

**Logistic:**

$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.

**Exponential:**

$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.

To find the PDF for these distributions we can take the derivative of the CDF. 

**PDF for Logistic Distribution:**

$$
f(x; \sigma) = \dfrac{e^{-x/\sigma}}{\sigma (1 + e^{−x/\sigma})^2}
$$
with $\sigma >0$.

**PDF for Exponential Distribution:**

$$
f(x; \lambda) = {\lambda}e^{-{\lambda}x}
$$
with $\lambda > 0$.

```python
import scipy as sp
from scipy.stats import logistic, expon, nbinom, rv_discrete
import matplotlib.pyplot as plt
import seaborn as sns

# --- Plot the PDF and CDF for various standard deviations ---
sd = [0.5, 1, 10]
x = np.linspace(-8, 8, 800)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# PDF plots
for s in sd:
    axes[0].plot(x, logistic.pdf(x, scale=s), label=f"$\\sigma$={s}")
axes[0].set_title("PDF of Logistic Distribution for Different Standard Deviations")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].legend()

# CDF plots
for s in sd:
    axes[1].plot(x, logistic.cdf(x, scale=s), label=f"$\\sigma$={s}")
axes[1].set_title("CDF of Logistic Distribution for Different Standard Deviations")
axes[1].set_xlabel("x")
axes[1].set_ylabel("F(x)")
axes[1].legend()

plt.tight_layout()
plt.show()

# --- Sample and compare KDE/ECDF with theoretical PDF/CDF ---
grid = np.linspace(-8, 8, 400)
pdf = logistic.pdf(grid)
cdf = logistic.cdf(grid)
sample = logistic.rvs(size=1000, random_state=100)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# KDE vs PDF
sns.kdeplot(x=sample, ax=axes[0], label='KDE (sample)')
sns.lineplot(x=grid, y=pdf, ax=axes[0], label='PDF (theoretical)', linestyle='--')
axes[0].set_title("Sample KDE vs Theoretical PDF")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Density")
axes[0].legend()

# ECDF vs CDF
sns.ecdfplot(x=sample, ax=axes[1], label='ECDF (sample)')
sns.lineplot(x=grid, y=cdf, ax=axes[1], label='CDF (theoretical)', linestyle='--')
axes[1].set_title("Sample ECDF vs Theoretical CDF")
axes[1].set_xlabel("x")
axes[1].set_ylabel("Cumulative Probability")
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# --- Plot the PDF and CDF for various scale parameters ---
scales = [0.5, 1, 2]   # scale = 1/λ
x = np.linspace(0, 10, 800)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# PDF plots
for s in scales:
    axes[0].plot(x, expon.pdf(x, scale=s), label=f"($\\lambda$={1/s:.2f})")
axes[0].set_title("PDF of Exponential Distribution for Different Scale Parameters")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].legend()

# CDF plots
for s in scales:
    axes[1].plot(x, expon.cdf(x, scale=s), label=f"($\\lambda$={1/s:.2f})")
axes[1].set_title("CDF of Exponential Distribution for Different Scale Parameters")
axes[1].set_xlabel("x")
axes[1].set_ylabel("F(x)")
axes[1].legend()

plt.tight_layout()
plt.show()

# --- Sample and compare KDE/ECDF with theoretical PDF/CDF ---
grid = np.linspace(0, 10, 400)
pdf = expon.pdf(grid, scale=1)   # default scale=1 (λ=1)
cdf = expon.cdf(grid, scale=1)
sample = expon.rvs(scale=1, size=1000, random_state=100)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# KDE vs PDF
sns.kdeplot(x=sample, ax=axes[0], label='KDE (sample)')
sns.lineplot(x=grid, y=pdf, ax=axes[0], label='PDF (theoretical)', linestyle='--')
axes[0].set_title("Sample KDE vs Theoretical PDF (Exponential)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Density")
axes[0].legend()

# ECDF vs CDF
sns.ecdfplot(x=sample, ax=axes[1], label='ECDF (sample)')
sns.lineplot(x=grid, y=cdf, ax=axes[1], label='CDF (theoretical)', linestyle='--')
axes[1].set_title("Sample ECDF vs Theoretical CDF (Exponential)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("Cumulative Probability")
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# --- Plot the PMF and CDF for various parameters ---
# n = number of successes, p = success probability
params = [(1, 0.1), (2, 0.2), (3, 0.3), (4, 0.4)]

# x range (discrete values)
x = np.arange(0, 40)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# PMF plots
for n, p in params:
    axes[0].stem(x, nbinom.pmf(x, n, p), basefmt=" ", label=f"r={n}, p={p}")
axes[0].set_title("PMF of Negative Binomial Distribution")
axes[0].set_xlabel("Number of Failures (x)")
axes[0].set_ylabel("P(X = x)")
axes[0].legend()

# CDF plots
for n, p in params:
    axes[1].plot(x, nbinom.cdf(x, n, p), marker='o', label=f"r={n}, p={p}")
axes[1].set_title("CDF of Negative Binomial Distribution")
axes[1].set_xlabel("Number of Failures (x)")
axes[1].set_ylabel("P(X ≤ x)")
axes[1].legend()

plt.tight_layout()
plt.show()

# --- Sample and compare sample histogram / ECDF with theoretical PMF / CDF ---
r, p = 4, 0.4  # choose one parameter setting for demonstration
sample = nbinom.rvs(r, p, size=1000, random_state=42)
x_vals = np.arange(0, np.max(sample) + 1)
pmf = nbinom.pmf(x_vals, r, p)
cdf = nbinom.cdf(x_vals, r, p)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram vs PMF
sns.histplot(sample, stat="probability", bins=np.arange(-0.5, np.max(sample) + 1.5), ax=axes[0], label='Sample histogram')
axes[0].stem(x_vals, pmf, basefmt=" ", linefmt='--', markerfmt='o', label='PMF (theoretical)')
axes[0].set_title("Sample Histogram vs Theoretical PMF (Negative Binomial)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Probability")
axes[0].legend()

# ECDF vs CDF
sns.ecdfplot(sample, ax=axes[1], label='ECDF (sample)')
sns.lineplot(x=x_vals, y=cdf, ax=axes[1], linestyle='--', label='CDF (theoretical)')
axes[1].set_title("Sample ECDF vs Theoretical CDF (Negative Binomial)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("Cumulative Probability")
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# --- Define several categorical distributions ---
categories = np.arange(1, 5)  # categories: 1, 2, 3, 4

# Different probability sets (must sum to 1)
probs = [
    [.1, .2, .3, .4],
    [.5, .2, .2, .1],
    [.8, .05, .05, .1]
]

# --- Plot PMF and CDF for different probability vectors ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# PMF
for i, p in enumerate(probs):
    dist = rv_discrete(name=f'cat{i}', values=(categories, p))
    axes[0].plot(categories, p, marker='o', label=f'Dist {i+1}')
axes[0].set_title("PMF of Categorical Distribution")
axes[0].set_xlabel("Category")
axes[0].set_ylabel("P(X = x)")
axes[0].set_xticks(categories)
axes[0].legend()

# CDF
for i, p in enumerate(probs):
    cdf = np.cumsum(p)
    axes[1].plot(categories, cdf, marker='o', label=f'Dist {i+1}')
axes[1].set_title("CDF of Categorical Distribution")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("P(X ≤ x)")
axes[1].set_xticks(categories)
axes[1].legend()

plt.tight_layout()
plt.show()

# --- Sample from one of them and compare sample frequencies / ECDF to theoretical ---
chosen_probs = [.1, .2, .3, .4]
dist = rv_discrete(name='cat', values=(categories, chosen_probs))
sample = dist.rvs(size=1000, random_state=42)

# Theoretical
pmf = chosen_probs
cdf = np.cumsum(pmf)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram vs PMF
sns.histplot(sample, stat="probability", bins=np.arange(0.5, len(categories) + 1.5), ax=axes[0], label='Sample histogram')
axes[0].plot(categories, pmf, 'o--', label='PMF (theoretical)')
axes[0].set_title("Sample Histogram vs Theoretical PMF (Categorical)")
axes[0].set_xlabel("Category")
axes[0].set_ylabel("Probability")
axes[0].set_xticks(categories)
axes[0].legend()

# ECDF vs CDF
sns.ecdfplot(sample, ax=axes[1], label='ECDF (sample)')
axes[1].plot(categories, cdf, 'o--', label='CDF (theoretical)')
axes[1].set_title("Sample ECDF vs Theoretical CDF (Categorical)")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("Cumulative Probability")
axes[1].set_xticks(categories)
axes[1].legend()

plt.tight_layout()
plt.show()
```
