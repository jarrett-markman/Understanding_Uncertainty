- The expected value of a single die roll is:

$$
\mathbb{E}[X] = \sum_{i=1}^{6} \frac{1}{6} x_i
$$

where $x_i$ represents each possible dice roll. 

We can write out the solution as: 

$$
\mathbb{E}[X] = \frac{1}{6}1+\frac{1}{6}2+\frac{1}{6}3+\frac{1}{6}4+\frac{1}{6}5+\frac{1}{6}6 = \frac{21}{6} = 3.5
$$

- The expected value of the addition of two dice is $7$ The math behind this is:

$$
\mathbb{E}[X] = 
\frac{1}{36}2 +
\frac{2}{36}3 +
\frac{3}{36}4 +
\frac{4}{36}5 +
\frac{5}{36}6 +
\frac{6}{36}7 +
\frac{5}{36}8 +
\frac{4}{36}9 +
\frac{3}{36}10 +
\frac{2}{36}11 +
\frac{1}{36}12 = \frac{252}{36} = 7
$$

- In European roulette, with probability $K/37$ you win and get $36/K$ dollars as payoff. There is a $(37-K)/37$ chance where you receive nothing back. Where $K$ represents the number of slots chosen. 

The expected value for a roulette game can be written as:

$$
\mathbb{E}[X] = (\frac{K}{37}) (\frac{36}{K}) + (\dfrac{37-K}{37})(0) -1
$$

The expected result comes from the likelihood of a win multiplied by the payoff plus the likelihood of a loss (0 payoff) and subtracting the bet you place which can be simplified to:

$$
\mathbb{E}[X] = \frac{36}{37} -1 \approx -0.027
$$

Instead of there being a $1/6$ probability of rolling a $6$, there is now a $1/6$ probability of rolling a $6$ and then also rolling any value from $1-6$. The equation then looks like:

$$
\mathbb{E}[X] = \dfrac{1}{6}1+\dfrac{1}{6}2+\dfrac{1}{6}3+\dfrac{1}{6}4+\dfrac{1}{6}5+\dfrac{1}{6}\left(6 + 3.5\right)

$$
\mathbb{E}[X] = \dfrac{21}{6} + \dfrac{1}{6}(3.5) = \dfrac{24.5}{6} \approx 4.083

$$
\mathbb{E}[X] = \frac{1}{6}1+\frac{1}{6}2+\frac{1}{6}3+\frac{1}{6}4+\frac{1}{6}5+\frac{1}{6}(6 + {\mathbb{E}[X]})

- Compute the variance for a uniform random variable

Based on our answer from question 1, we know that $\mathbb{E}[X] = 0.5$.

Therefore, we can use an integral to calculate the variance:

$$
\mathbb{V}[X] = \int_0^1 (x-.5)^2 dx = \int_0^1 (x^2 - x + .25) dx 
$$

After integrating, we can use the integral to calculate the value for the variance as:

$$
\mathbb{V}[X] = \left[ \frac{1}{3} x^3 - \frac{1}{2}x^2 + .25x \right]_{0}^{1} = \frac{1}{3}-\frac{1}{2}+\frac{1}{4} = \frac{1}{12}
$$

- Show that $\mathbb{V}[X] = \mathbb{E}[X^2] - \mathbb{E}[X]^2$

We know that: $\mathbb{V}[X] = \int (x-\mathbb{E}[X])^2 f(x) dx$

Therefore, we can solve this integral:

$$
= \int (x^2 - 2 x \mathbb{E}[X] + \mathbb{E}[X]^2 ) f(x) dx \\
= \int x^2 f(x)dx - 2 \int xf(x)dx \mathbb{E}[X] + \mathbb{E}[X]^2 \\
= \mathbb{E}[X^2] - 2 \mathbb{E}[X]^2 + \mathbb{E}[X]^2 = \mathbb{E}[X^2] - \mathbb{E}[X]^2 
$$

- Show that $\mathbb{V}[a+bX] = b^2 \mathbb{V}[X]$

We know that: $\mathbb{V}[a+bX] = \int (a+bX - \mathbb{E}[a+bX])^2 f(x)dx$

Like with the prior part, we can solve the integral:


$$
= \int (a+bX - a - b \mathbb{E}[X])^2 f(x)dx \\
= \int (bX - b \mathbb{E}[X])^2 f(x)dx \\
= \int b^2(X - \mathbb{E}[X])^2 f(x)dx \\
=  b^2\int(X - \mathbb{E}[X])^2 f(x)dx \\
=  b^2 \mathbb{V}[X]
$$

- Show that if $X$ is a normally distributed random variable, then $a + bX$ is distributed normally with mean $a+ b \mathbb{E}[X]$ and variance $b^2 \sigma_X^2$ 

We know that $X$ is normally distributed with mean $\mu$, variance $\sigma_X^2$ and $Y = a + b X$. Then $X = (Y-a)/b$. 

We can start with the distribution function for X:

$$
F_X\left( \frac{Y-a}{b} \right) = \int_{-\infty}^{+\infty} \frac{1}{\sqrt{2 \pi}} \exp \left\lbrace - \frac{1}{2} \left( \dfrac{(Y-a)/b - \mu}{\sigma} \right)^2  \right\rbrace dx =  \int_{-\infty}^{+\infty} \frac{1}{\sqrt{2 \pi}} \exp \left\lbrace - \frac{1}{2} \left( \dfrac{Y-a - b\mu}{b\sigma} \right)^2  \right\rbrace dx
$$

This is the same form of a normal distribution with mean $a + b\mu = \mu'$ and standard deviation $b\sigma = \sigma'$ which we can write as:
$$
F_Y(Y) = \int_{-\infty}^{+\infty} \frac{1}{\sqrt{2 \pi}} \exp \left\lbrace - \frac{1}{2} \left( \dfrac{Y-\mu'}{\sigma'} \right)^2  \right\rbrace dx
$$

Showing that $Y$ is distributed normally with mean $\mathbb{E}[Y] = a + b \mathbb{E}[X]$ and standard deviation $\sigma_Y' = b \sigma_X$

- The **covariance** of $X$ and $Y$ is
$$
\text{cov}(X,Y) = \int_{y} \int_{x} (x-\mathbb{E}[X])(y-\mathbb{E}[Y])f_{XY}(x,y) dxdy = \mathbb{E}_{XY}[ (x-\mu_X)(y-\mu_Y)]
$$
- Show that if $f_{XY}(x,y)=f_X(x)f_Y(y)$, then $\text{cov}(X,Y)=0$
- Provide an example (computation/simulation is fine) where $\text{cov}(X,Y)\approx 0$ but $f_{XY}(x,y)\neq 0$
- The covariance doesn't characterize joint random variables except in a few special cases: The covariance only captures the **linear** association between the two variables, not nonlinear associations.

- Show that if $f_{XY}(x,y)=f_X(x)f_Y(y)$, then $\text{cov}(X,Y)=0$

We know that the **covariance of $X$ and $Y$** is:
$$
\text{cov}(X,Y) = \int_{y} \int_{x} (x-\mathbb{E}[X])(y-\mathbb{E}[Y])f_{XY}(x,y) dxdy = \mathbb{E}_{XY}[ (x-\mu_X)(y-\mu_Y)]
$$

Through integration we can show that $\text{cov}(X,Y)=0$:

$$
\text{cov}(X,Y) = \int_{y} \int_{x} (x-\mathbb{E}[X])(y-\mathbb{E}[Y])f_{X}(x)F_y(y) dxdy \\
= \int_{y} \left[ \int_{x} (x-\mathbb{E}[X])f_{X}(x) dx \right] (y-\mathbb{E}[Y])F_y(y)dy \\
= \int_{y} \left[ \int_{x} xf_{X}(x) dx-\mathbb{E}[X] \right] (y-\mathbb{E}[Y])F_y(y)dy \\
= \int_{y} \left[ \mathbb{E}[X]-\mathbb{E}[X] \right] (y-\mathbb{E}[Y])F_y(y)dy \\
= \int_{y} \left[ 0 \right] (y-\mathbb{E}[Y])F_y(y)dy \\
= 0
$$

When there is non-linear dependence between the variables, it's possible to have a join density function where the covariance is zero, but the density is not zero. 
An example of this could be the uniform distribution:
- X ~ Uniform(-1, 1) and Y ~ Uniform(-1, 1)

The covariance in this case is:

$$
\text{Cov}(X, Y) = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y] = 0 - 0 \times 0 = 0
$$

However, the join density in this case is:

$$
f_{xy}(x, y) = \begin{cases}
\frac{1}{4}, & -1 < x < 1, \ -1 < y < 1,\\[2mm]
0, & \text{otherwise.}
\end{cases}
$$

- Let treatment, $T$ be distributed binomial with parameter $p$. Then the contribution to the likelihood for each patient $i$, with $y_i = 0$ for no radiation therapy and $y_i=1$ for radiation therapy, is 
$$
p^{y_i}(1-p)^{1-y_i}
$$
- Write out the likelihood.
- Maximize the likelihood with respect to $p$. What is the MLE, $\hat{p}$?
- Bootstrap the sampling density/distribution of $\hat{p}$.

```python
# Import pandas and scipy
import pandas as pd
import scipy as sp 

# Import metabric data
data = pd.read_csv('metabric.csv')

# Plot histogram for Radio Therapy
sns.histplot(data['Radio Therapy'], stat='proportion')
```

The likelihood is:
$$
L(p) = \prod_{i=1}^n p^{y_i}(1-p)^{1-y_i}
$$

To maximize, we first take logs:

$$
\ell(p) = \sum_{i=1}^n y_i \log(p) + (1-y_i) \log(1-p)
$$

Next the derivative of the equation:

$$
\ell'(p) = \sum_{i=1}^n y_i \frac{1}{p} - (1-y_i) \frac{1}{1-p}
$$

Where at the optimum, we have:

$$
\sum_{i=1}^n y_i \frac{1}{\hat{p}} - (1-y_i) \frac{1}{1-\hat{p}} = 0
$$

Solving the first-order condition for $p$ yields the MLE

$$
\hat{p} = \dfrac{1}{n} \sum_{i=1}^n y_i.
$$

```python
# Bootstrap density/distribution of p hat

y = pd.to_numeric( data['Radio Therapy'].replace({'NO':'0','YES':'1'}) , errors='coerce' ) # Convert string to numeric
print(f'MLE: ', y.mean())

S = 1000 # 1000 bootstrap samples
sample = [np.mean( y.sample(frac=1,replace=True)) for s in range(S) ]

fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x=sample,ax=axes[0],label='kde')
axes[0].set_title('Sample KDE')
sns.ecdfplot(x=sample,ax=axes[1],label='ecdf')
axes[1].set_title('Sample ECDF')
plt.show()
```
